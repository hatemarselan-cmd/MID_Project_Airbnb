### 1) Split Data into Input & Output

```python
x = df.drop('log_price', axis= 1) ## saving df without log_price 
y = df['log_price'] ## target Feature
```

### 2) Split Data into Train & Test

```python
from sklearn.model_selection import train_test_split 

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.2, random_state= 42)### test_size=0.2 → 20% of the data goes to testing
```

### 3) Handle Numerical

## Missing Values Handling Plan

Below is the recommended action for each column based on the percentage of missing values.

### 🟠 Columns to FILL (missing between 5% and 40%)
| Column              | Missing % | Action | Suggested Strategy |
|---------------------|-----------|--------|---------------------|
| host_response_rate   | 24.69%   | **Fill** | after deep diving i  decided  KNN |
| review_scores_rating | 22.56%   | **Fill** | after deep diving i  decided  KNN |

### 🟢 Columns to DROP ROWS (missing < 5%)
| Column                  | Missing % | Action |
|-------------------------|-----------|--------|
| bathrooms               | 0.27%     | **Drop rows** | 
| host_experience_years   | 0.25%     | **Drop rows** |
| host_identity_verified  | 0.25%     | **Drop rows** | 
| host_has_profile_pic    | 0.25%     | **Drop rows** |
| beds                    | 0.18%     | **Drop rows** | 
| bedrooms                | 0.12%     | **Drop rows** |


```python
## drop  rows  for columns <5%

cols_to_drop_na = ['bathrooms', 'host_experience_years', 'host_identity_verified', 'host_has_profile_pic', 'beds', 'bedrooms']
x_train = x_train.dropna(subset=cols_to_drop_na).reset_index(drop=True)
x_test = x_test.dropna(subset=cols_to_drop_na).reset_index(drop=True)
```

```python
x_train['host_has_profile_pic']=x_train['host_has_profile_pic'].astype(bool)
x_test['host_has_profile_pic']=x_test['host_has_profile_pic'].astype(bool)
x_train['host_identity_verified']=x_train['host_identity_verified'].astype(bool)
x_test['host_identity_verified']=x_test['host_identity_verified'].astype(bool)
```

```python
x_train['host_has_profile_pic'].dtypes
```

#### 3.1) Impute Missing Values

```python
num_cols = x_train.select_dtypes(include= 'number').columns
num_cols
```

```python
x_train[num_cols].isna().mean().round(4).sort_values(ascending=False)*100
```

```python
x_test[num_cols].isna().mean().round(4).sort_values(ascending=False)*100
```

```python
# host_response_rate , review_scores_rating --> Fill with Knn Imputer
from sklearn.impute import KNNImputer

knn = KNNImputer()

x_train[num_cols] = knn.fit_transform(x_train[num_cols])
x_test[num_cols] = knn.transform(x_test[num_cols])
```

```python
x_train.isna().mean().round(4).sort_values(ascending=False)*100## to check that there is no missing value any more 
```

```python
x_test.isna().mean().round(4).sort_values(ascending=False)*100## to check that there is no missing value any more 
```

#### 3.2) Feature Scaling

```python
# Use Robust Scaler as it is not affected by outliers
from sklearn.preprocessing import RobustScaler

robust_scaler = RobustScaler()

x_train[num_cols] = robust_scaler.fit_transform(x_train[num_cols])
x_test[num_cols] = robust_scaler.transform(x_test[num_cols])
```

#### 4.1) Encoding Ordinal Columns

```python
# cancellation_policy  encoding with ordinal encoder
from sklearn.preprocessing import OrdinalEncoder

ord_encoder = OrdinalEncoder(categories= [ ['super_strict_60', 'super_strict_30','strict', 'moderate', 'flexible'] ])

x_train[['cancellation_policy']] = ord_encoder.fit_transform(x_train[['cancellation_policy']])
x_test[['cancellation_policy']] = ord_encoder.transform(x_test[['cancellation_policy']])
```

```python
x_train['cancellation_policy'].head()
```

#### 4.3) Encoding Nominal (OneHotEncoder)

```python
nominal_cols = x_train.select_dtypes(include= 'object').columns
nominal_cols
```

```python
from sklearn.preprocessing import OneHotEncoder## i will use onehotencoder as the nominal feature less than 7 variables 

ohe = OneHotEncoder(drop= 'first', sparse_output= False)

ohe_arr_train = ohe.fit_transform(x_train[nominal_cols])
ohe_arr_test = ohe.transform(x_test[nominal_cols])
```

```python
## creating dataframe from the result from onhotencoder 
ohe_df_train = pd.DataFrame(ohe_arr_train, columns = ohe.get_feature_names_out())
ohe_df_test = pd.DataFrame(ohe_arr_test, columns = ohe.get_feature_names_out())
```

```python
# Reset index before concatenation
x_train.reset_index(drop= True, inplace= True)
x_test.reset_index(drop= True, inplace= True)

y_train.reset_index(drop= True, inplace= True)
y_test.reset_index(drop= True, inplace= True)
```

```python
## concatenate the output to x_train , x_test then drop orginal columns 
x_train = pd.concat([x_train, ohe_df_train], axis= 1).drop(columns= nominal_cols, axis= 1)
x_test = pd.concat([x_test, ohe_df_test], axis= 1).drop(columns= nominal_cols, axis= 1)
```
